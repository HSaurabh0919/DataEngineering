```python
### Working with Pyspark
from pyspark.sql import SparkSession
```

```python
# Create Spark session
spark = SparkSession.builder \
    .appName("EmployeeData") \
    .getOrCreate()
```

```python
# Sample employee data
employee_data = [
    (101, "Saurabh", "Data Science", 85000, "Hyderabad"),
    (102, "Ankit", "Backend", 72000, "Bangalore"),
    (103, "Priya", "HR", 50000, "Pune"),
    (104, "Rahul", "DevOps", 78000, "Chennai"),
    (105, "Sneha", "Data Science", 68000, "Delhi")
]
columns = ["emp_id", "name", "department", "salary", "city"]
```

```python
# Create DataFrame
df = spark.createDataFrame(employee_data, columns)
```

```python
df.show()
```

### Operations of Spark dataframe

```python
###Filtering operations
df1 = df.filter(
    (df.salary > 60000) & (df.city != 'Delhi')
)

df1.show()
```

```python
from pyspark.sql.functions import count, mean
df.groupby("department") \
.agg(mean("salary").alias("salary_mean")) \
.show()
```

```python
from pyspark.sql.functions import avg

# Compute mean salary for each department
dept_avg_salary = df.groupBy("department") \
    .agg(avg("salary").alias("avg_salary"))

# Join back to original dataframe
df_final = df.join(dept_avg_salary, on="department", how="left")

df_final.show()
```
